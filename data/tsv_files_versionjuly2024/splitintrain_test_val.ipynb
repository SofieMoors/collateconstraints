```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Step 1: Read the CSV file into a DataFrame
file_path = '/Users/sofiemoors/Library/CloudStorage/OneDrive-UniversiteitAntwerpen/Documenten/Vormvast/collate_constraints/data/tsv_files_corrected/tsv_inputretrain.tsv'
# Update this with the path to your CSV file
data = pd.read_csv(file_path, delimiter='\t', header=0)
data.dropna(how='all', inplace=True)

# Step 2: Split into 80% train and 20% combined validation + test
train_data, temp_data = train_test_split(data, test_size=0.2, random_state=42, shuffle=False)

# Step 3: Split the combined 20% into 10% validation and 10% test
val_data, test_data = train_test_split(temp_data, test_size=0.5, random_state=42, shuffle=False)

# Print the lengths of each set to verify the splits
print(f"Length of train_data: {len(train_data)}")
print(f"Length of val_data: {len(val_data)}")
print(f"Length of test_data: {len(test_data)}")
```

```python
# Save the train, validation, and test sets to CSV files

train_data.to_csv('/Users/sofiemoors/Library/CloudStorage/OneDrive-UniversiteitAntwerpen/Bureaublad/train_data.tsv',index=False, sep='\t')
val_data.to_csv('/Users/sofiemoors/Library/CloudStorage/OneDrive-UniversiteitAntwerpen/Bureaublad/val_data.tsv', index=False, sep='\t')
test_data.to_csv('/Users/sofiemoors/Library/CloudStorage/OneDrive-UniversiteitAntwerpen/Bureaublad/test_data.tsv', index=False, sep='\t')

print("Data has been saved to CSV files.")
```
